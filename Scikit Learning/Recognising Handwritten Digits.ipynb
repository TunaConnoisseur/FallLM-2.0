```python
# Imports
import matplotlib.pyplot as plt

from sklearn import datasets, metrics, svm
from sklearn.model_selection import train_test_split

digits = datasets.load_digits()

# Create a figure with 10 subplots
_, axes = plt.subplots(nrows=1, ncols=10, figsize=(12, 10))
# Merges the axes and images together to display the images of the digits along with their labels
for ax, image, label in zip(axes, digits.images, digits.target):
    ax.set_axis_off()
    # Displays the image of the digit, in grayscale
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation="nearest")
    ax.set_title("Training: %i" % label)
```

```python
# Flatten images
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

# Create classifier: a support vector classifier
classifier = svm.SVC(gamma=0.001)

# Split data into 50% train and 50% test subsets
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.5, shuffle=False
)

# Learn the digits on the train subset
classifier.fit(X_train, y_train)

predicted = classifier.predict(X_test)
```

```python
_, axes = plt.subplots(nrows=1, ncols=10, figsize=(15, 10))
for ax, image, prediction in zip(axes, X_test, predicted):
    ax.set_axis_off()
    image=image.reshape(8, 8)
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation="nearest")
    ax.set_title(f"Prediction: {prediction}")
```

```python
print(
    f"Classification report for classifier {classifier}:\n"
    f"{metrics.classification_report(y_test, predicted)}"
)
```

```python
disp = metrics.ConfusionMatrixDisplay.from_predictions(y_test, predicted)
disp.figure_.suptitle("Confusion Matrix")
print(f"Confusion matrix:\n{disp.confusion_matrix}")

plt.show()
```
